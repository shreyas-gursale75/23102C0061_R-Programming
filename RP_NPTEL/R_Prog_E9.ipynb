```r
dates = c("2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27")
```

```r
datesplt = strsplit(dates, "-")
print(datesplt)
```

```r
datemat = matrix(unlist(datesplt), nrow = 4, ncol=3, byrow=TRUE)
print(datemat)
```

```r
datematrix = matrix(as.numeric(unlist(datesplt)), nrow = 4, ncol=3, byrow=TRUE)
print(datematrix)
```
