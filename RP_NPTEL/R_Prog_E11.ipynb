```r
# Load MASS package and view the painters data frame
library(MASS)

painters
```

```r
# Summary of categorical and numeric variables
summary(painters$School)
summary(painters$Composition)
```

```r
# Attach the data frame so variables can be referenced directly
attach(painters)

summary(School)
summary(Composition)

# Recover the default setting
detach(painters)
```

```r
# Subset painters by School
subset(painters, School == "F")

# Equivalent indexing approach
painters[painters[["School"]] == "F", ]
```

```r
# Subset using a numeric condition
subset(painters, Composition <= 6)
```

```r
# Remove uninteresting columns
# Removes column 3 (Colour) and column 5 (School)
subset(painters, School == "F", select = c(-3, -5))
```

```r
# Split the data frame according to School
splitted = split(painters, painters$School)

splitted

# Each element is itself a data frame
is.data.frame(splitted$A)
```

```r
# Heights of 50 persons
height = c(
  166,125,130,142,147,159,159,147,165,156,
  149,164,137,166,135,142,133,136,127,143,
  165,121,142,148,158,146,154,157,124,125,
  158,159,164,143,154,152,141,164,131,152,
  152,161,143,143,139,131,125,145,140,163
)

# Scatter plot of one variable
plot(height)
```

```r
# Gender data
gender = c(1, 2, 1, 2, 1, 1, 1, 2, 1, 1)

# Direct plot (not frequency-based)
barplot(gender)

# Frequency table
table(gender)

# Bar plot using absolute frequencies
barplot(table(gender))
```

```r
# Bar plot using relative frequencies
barplot(table(gender) / length(gender))
```

```r
# Add labels and colours
barplot(
  table(gender),
  names.arg = c("Male", "Female"),
  col = c("lightblue", "pink"),
  main = "Gender Distribution",
  xlab = "Gender",
  ylab = "Frequency"
)
```

```r
# Pie chart using gender frequencies
pie(
  table(gender),
  main = "Gender Distribution",
  col = c("lightblue", "pink")
)
```

```r
# Histogram of heights
hist(height)
```

```r
# Styled histogram
hist(
  height,
  main = "Heights of Persons",
  col = "green",
  xlab = "Heights",
  ylab = "Number of Persons"
)
```

```r
# Display a bar plot and pie chart together
par(mfrow = c(1, 2))

barplot(table(gender))
pie(table(gender))

# Restore one-plot layout
par(mfrow = c(1, 1))
```
