## Knowing numbers and characters

```python
%%R
x=20

print(is.numeric(x))

print(is.character(x))
```

```python
%%R
m="apple"

print(is.numeric(m))

print(is.character(m))
```

## Converting numbers and characters

```python
%%R
x = 20
print(is.numeric(x))

y = as.character(x)
print(is.numeric(y))

print(is.character(y))

print(y)
```

```python
%%R
y = "apple"
print(is.numeric(y))
print(is.character(y))

z = suppressWarnings(as.numeric(y))
print(is.numeric(z))

print(is.character(z))

print(z)
```

## Matrix

```python
%%R
x = matrix( nrow=4, ncol=2,data=c(1,2,3,4,5,6,7,8) )

print(x)

print(x[3,2])
```

## Entering data column wise in a matrix

```python
%%R
x = matrix( nrow=4, ncol=2,data=c(1,2,3,4,5,6,7,8))
print(x)
```

```python
%%R
x = matrix( nrow=4, ncol=2,data=c(1,2,3,4,5,6,7,8), byrow = FALSE)
print(x)
```

## Entering data row wise in a matrix

```python
%%R
y = matrix( nrow=4, ncol=2,data=c(1,2,3,4,5,6,7,8), byrow = TRUE)

print(y)
```
